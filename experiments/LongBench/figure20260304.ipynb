```python
import matplotlib.pyplot as plt
import numpy as np

# ---------------- 数据 ----------------
capacity = np.array([512, 1024, 2048, 4096])

baseline = np.array([28.78, 29.57, 32.44, 33.71])
sliding = np.array([15.15, 17.63, 22.21, 29.49])
sliding_comp = np.array([15.31, 17.3, 22.11, 29.24])
snapkv_cam = np.array([27.82, 29.54, 32.39, 32.88])

# ---------------- 绘图 ----------------
plt.figure(figsize=(8,6))

plt.plot(capacity, baseline, marker='o', linewidth=3, linestyle='--', label='Baseline (No Compensation)')
plt.plot(capacity, sliding, marker='s', linewidth=2, label='Sliding Window')
plt.plot(capacity, sliding_comp, marker='^', linewidth=2, label='Sliding Window + Compensation')
plt.plot(capacity, snapkv_cam, marker='d', linewidth=2, label='SnapKV + CaM')

# ---------------- 细节设置 ----------------
plt.xlabel("Max KV Cache Capacity", fontsize=12)
plt.ylabel("LongBench Score", fontsize=12)

plt.xticks(capacity)
plt.grid(True, linestyle='--', alpha=0.5)

plt.legend()
plt.tight_layout()

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# ---------------- 数据 ----------------
capacity = np.array([512, 1024, 2048, 4096])

baseline = np.array([28.78, 29.57, 32.44, 33.71])
snapkv_math = np.array([27.79, 29.57, 32.44, 33.13])

# ---------------- 绘图 ----------------
plt.figure(figsize=(8,6))

# 对照组
plt.plot(
    capacity,
    baseline,
    marker='o',
    linestyle='--',
    linewidth=3,
    color='black',
    label='Baseline (No Compensation)'
)

# SnapKV + Math
plt.plot(
    capacity,
    snapkv_math,
    marker='s',
    linewidth=2.5,
    color='#1f77b4',
    label='SnapKV + Math'
)

# ---------------- 细节 ----------------
plt.xlabel("Max KV Cache Capacity", fontsize=12)
plt.ylabel("LongBench Score", fontsize=12)

plt.xticks(capacity)

plt.grid(True, linestyle='--', alpha=0.5)

plt.legend()
plt.tight_layout()

plt.show()
```
